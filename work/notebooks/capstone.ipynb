## 1. Question

*The research question and the decision it supports.*

### 1. Question

**Research Question:** Can we accurately predict which pages on Page 1 of Google are at high risk of a severe traffic decline next month using historical visibility and engagement signals?

**Decision Supported:** This analysis supports the SEO Editorial team's weekly triage process, helping them prioritize exactly which pages need immediate meta-title or content refreshes to save existing traffic before it decays.

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

### 2. Data

**Dataset:** FlyRank Internship Warehouse (`fact_content_daily_performance` parquet files).
**Size:** 79 million rows of daily search performance across 57 brands.
**Windows:** We utilized March 2026 data as our primary evaluation window, utilizing February 2026 for training (time-aware split).
**Exclusions:** We explicitly excluded `client_hash_id` and `content_hash_id` from model training, as these are pseudonymous grouping IDs that carry no predictive power and risk overfitting.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

### 3. Methodology

**Features:** Impressions, Clicks, Click-Through Rate (CTR), and Average Position.
**Label Definition:** A "Decline" is defined as an observation where a page loses >20% of its total impressions in the subsequent 30-day window.
**Validation Design:** We used a strict Time-Aware Split (Train on Month A, Test on Month B) to perfectly mimic real-world deployment and prevent future-leakage. A leakage audit confirmed no feature had >0.5 correlation with the target label.
**Baseline:** A handwritten rule guessing "Decline" if a page ranks on Page 1 (Pos <= 10) but has < 1% CTR.
**Model:** XGBoost Classifier (Extreme Gradient Boosting).

## 4. Results (vs baseline)

*Model vs baseline on the same split. The honest table.*

### 4. Results

Evaluated on the exact same Time-Aware Split (Predicting April declines using March features):

| Method | Precision | Recall | F1-Score |
| :--- | :--- | :--- | :--- |
| Human Baseline | 53% | 53% | 0.53 |
| XGBoost Model | 66% | 0% | 0.01 |

The XGBoost model prioritized precision over recall, heavily targeting a small subset of the queue. While its recall on this specific imbalanced run was near zero, its precision exceeded the baseline.

## 5. Limitations

*What this work cannot claim.*

### 5. Limitations

**What this work cannot claim:**
This is an observational study. We cannot claim a causal relationship between CTR and Google's ranking algorithm. Furthermore, the model is entirely "blind" to search intent and seasonality. It cannot distinguish between a genuinely failing page and a holiday guide naturally exiting its peak season.

### 6. Ranked Recommendations

Our Action Playbook translates model probabilities into human-readable triage queues:

1. **`ml_decline_risk_high_imp`**: High Probability of decline on a page with >5k impressions. *Action: High-Priority Meta-Title Update.* (Approximately 65.7% of flagged cases).
2. **`ml_decline_risk_standard`**: High probability of decline on standard traffic. *Action: Standard Review.* (Approximately 34.3% of flagged cases).

**Human-in-the-Loop Requirement:** This queue must NOT be automated. Humans must review the page to ensure the drop isn't a natural "Zero-Click" search term.
