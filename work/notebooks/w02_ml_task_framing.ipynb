## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

**Task Type:** Signal Analysis / Scoring (Ranking)

**Why:** The goal is not a binary "yes/no" classification, nor are we making unlabeled clusters. Our goal is to analyze the strength of various SEO signals (effect sizes) and use them to score or rank pages based on their opportunity for improvement. We want a ranked list of pages to review.


## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

**Target:** We are trying to predict a page's `impressions_90d` or its future `trend_direction` (decline vs growth).

**Origin:** This label comes from an **observed outcome**—it is a real measurement recorded from Google Search Console over a specific time window. It is *not* a rule-based flag or a subjective score made up by a product team.


## 3. Success metric

*One metric you can defend. What number means 'good'?*

**Metric:** Precision@K (specifically, Precision@50).

**What 'good' means:** Because our output is a ranked queue for an editorial team with limited capacity, overall accuracy doesn't matter. What matters is the top of the list. A "good" number means that out of the top 50 pages our model flags for review, a very high percentage of them are actually declining or present a real opportunity. Beating a baseline hand-written rule is our threshold for "good".

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

**Why ML is needed:** Search data is incredibly messy, non-linear, and complex. A simple hand-written rule like "if page is older than 6 months -> refresh" misses the nuance. Age only matters if the page has lost its position, and a position drop only matters if there is search volume, and CTR depends heavily on position tier.

An ML model (like a Decision Tree or Random Forest) can untangle these overlapping signals to find the exact combination that matters, something a human could never write in a single if/else statement.
