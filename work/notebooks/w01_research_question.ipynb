## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load the starter dataset
df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

# 1. Check if search volume correlates with actual traffic 
vol_corr = df["search_volume"].corr(df["impressions_90d"])

# 2. Check if word count correlates with average position
word_corr = df["word_count"].corr(df["avg_position"])

print(f"Correlation between search volume and traffic: {vol_corr:.3f}")
print("Result: Near zero. Search volume does not guarantee traffic.")
print(f"\nCorrelation between word count and position: {word_corr:.3f}")
print("Result: Very weak. Just writing more words does not guarantee a better ranking.")
print("\nConclusion: We cannot rely on assumptions. We need a deeper signal analysis!")
```
